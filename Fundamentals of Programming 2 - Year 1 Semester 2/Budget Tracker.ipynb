## Documentation

### budget_tracker.py

This python file contains all of the needed functions for the app.

It firstly imports csv since it's in a csv file where all of the transactions are stored.

The add() function asks the user for all necessary inputs and then appends all to the transactions.csv file. It checks each input to see if the ranges and input types are correct. It then returns the row for the testing function test_add() (in the testing python file test_budget_tracker.py) to check if the function add() works as intended.

The view() function opens the file in read mode and prints each row of the file. It also returns all rows in a list format for the test_view() function.

The update() function asks the user for the description of the transaction that the user wants to update. The function reads the csv file and locates the row with the chosen description, it then asks the user for a new amount for that transaction and updates that row. After, an empty list called 'new_rows' appends all of the rows existing in the csv file (including the updated row). Then there's an 'exist' variable that is either true or false. If it is false, that means the chosen description does not exist in any transaction and nothing has changed. If it is true however, then the file goes in write mode and rewrites down each row from the 'new_rows' list (which contains the updated row). It finally returns all new rows for the test_update() function.

The delete() function has the same structure to the update() function, with the exception being that when the file is in read mode, the function firstly checks what rows do NOT match the chosen description, and appends those rows to the 'new_rows' list. This excludes the chosen row, therefore deleting it when rewriting all of the contents of the 'new_rows' list into the csv file. It finally returns all new rows for the test_delete() function.

The search() function also has a similar structure to the update() function and also asks for the description of the chosen transaction, then in read mode, it searches for the row and prints it. If no description was found, no row is printed. It also returns the specific row for the test_search() function.

### test_budget_tracker.py

This python file firstly imports sys to get the correct file path to use the budget_tracker.py file to import all of the functions. It also imports csv for the file, and patch from unittest.mock for the testing.

Then it empties the transactions.csv file to do the testing of functions easily (it's easier to start with an empty csv file as I test each function).

The test_add() function simulates user input for all of the elements in a transaction. Then the function uses assert to verify if the output of the function add(), being a newly created row, is the expected one.

The test_view() function simply uses the assert to see if all of the added transactions are indeed listed.

The test_update() function also simulates user input to update the 'shop' transaction with a new amount value, and then also uses assert to verify if the correct row has been modified correctly.

The test_delete() function also simulates input and deletes the 'shop' transaction and uses assert to verify if the correct transaction has been deleted.

The test_search() function also simulates input and prints the wanted transaction, using assert to verify if the correct transaction has been displayed.

Lastly, it checks if all test functions have run correctly and prints "All tests passed." if so.

### main.py

This python file starts by importing the functions from the budget_tracker.py file to use them in the main() function. It also imports csv.

The file then writes onto the transactions.csv file the headers ("Date", "Description", "Amount", and "Category").

Then the main() function shows the user the available options for the app (like a manu) and then asks the user for their  choice. And once chosen, the function executes the associated function (add, view, update, delete, or search). The user can also exit the menu, and if an invalid action is inputted by the user, the function prompts the user to try again.
